# Cancel All Opened / Queued SageMaker Jobs

This notebook stops all **opened (InProgress)** or **queued (Stopping)** SageMaker jobs:
- Training Jobs
- Processing Jobs
- AutoML Jobs
- Transform Jobs

Run each cell to cancel the corresponding job type.

```python
import boto3

region = boto3.Session().region_name
sm_client = boto3.client('sagemaker', region_name=region)

print(f"Using region: {region}")
```

```python
# Stop all InProgress and Stopping Training Jobs
print("=== Stopping Training Jobs ===")
for status in ['InProgress', 'Stopping']:
    paginator = sm_client.get_paginator('list_training_jobs')
    for page in paginator.paginate(StatusEquals=status):
        for job in page['TrainingJobSummaries']:
            job_name = job['TrainingJobName']
            print(f"  Stopping training job: {job_name} (status: {status})")
            try:
                sm_client.stop_training_job(TrainingJobName=job_name)
                print(f"    -> Stopped: {job_name}")
            except sm_client.exceptions.ResourceNotFound:
                print(f"    -> Not found (already stopped?): {job_name}")
            except Exception as e:
                print(f"    -> Error stopping {job_name}: {e}")

print("Done with Training Jobs.")
```

```python
# Stop all InProgress and Stopping Processing Jobs
print("=== Stopping Processing Jobs ===")
for status in ['InProgress', 'Stopping']:
    paginator = sm_client.get_paginator('list_processing_jobs')
    for page in paginator.paginate(StatusEquals=status):
        for job in page['ProcessingJobSummaries']:
            job_name = job['ProcessingJobName']
            print(f"  Stopping processing job: {job_name} (status: {status})")
            try:
                sm_client.stop_processing_job(ProcessingJobName=job_name)
                print(f"    -> Stopped: {job_name}")
            except sm_client.exceptions.ResourceNotFound:
                print(f"    -> Not found (already stopped?): {job_name}")
            except Exception as e:
                print(f"    -> Error stopping {job_name}: {e}")

print("Done with Processing Jobs.")
```

```python
# Stop all InProgress Transform Jobs
print("=== Stopping Transform Jobs ===")
paginator = sm_client.get_paginator('list_transform_jobs')
for page in paginator.paginate(StatusEquals='InProgress'):
    for job in page['TransformJobSummaries']:
        job_name = job['TransformJobName']
        print(f"  Stopping transform job: {job_name}")
        try:
            sm_client.stop_transform_job(TransformJobName=job_name)
            print(f"    -> Stopped: {job_name}")
        except sm_client.exceptions.ResourceNotFound:
            print(f"    -> Not found (already stopped?): {job_name}")
        except Exception as e:
            print(f"    -> Error stopping {job_name}: {e}")

print("Done with Transform Jobs.")
```

```python
# Stop all InProgress and Stopping AutoML Jobs
print("=== Stopping AutoML Jobs ===")
for status in ['InProgress', 'Stopping']:
    paginator = sm_client.get_paginator('list_auto_ml_jobs')
    for page in paginator.paginate(StatusEquals=status):
        for job in page['AutoMLJobSummaries']:
            job_name = job['AutoMLJobName']
            print(f"  Stopping AutoML job: {job_name} (status: {status})")
            try:
                sm_client.stop_auto_ml_job(AutoMLJobName=job_name)
                print(f"    -> Stopped: {job_name}")
            except sm_client.exceptions.ResourceNotFound:
                print(f"    -> Not found (already stopped?): {job_name}")
            except Exception as e:
                print(f"    -> Error stopping {job_name}: {e}")

print("Done with AutoML Jobs.")
```
