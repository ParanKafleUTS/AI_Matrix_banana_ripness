```python
import boto3

# Your S3 bucket and prefix
bucket = "sagemaker-ap-southeast-2-443142193439"
prefix = "banana-ripeness-dataset/"

# Create S3 client
s3 = boto3.client("s3")

# List objects with the given prefix
response = s3.list_objects_v2(
    Bucket=bucket,
    Prefix=prefix,
    Delimiter="/"  # This makes S3 return "folders"
)

# Print all folder names
print("Folders inside", prefix)
if "CommonPrefixes" in response:
    for folder in response["CommonPrefixes"]:
        print(folder["Prefix"])
else:
    print("No folders found")
```

```python
prefix = "banana-ripeness-dataset/valid/"

response = s3.list_objects_v2(
    Bucket=bucket,
    Prefix=prefix,
    Delimiter="/"
)

print("Subfolders inside", prefix)
if "CommonPrefixes" in response:
    for folder in response["CommonPrefixes"]:
        print(folder["Prefix"])
```

```python
import boto3

# S3 bucket and the folder you want to check
bucket = "sagemaker-ap-southeast-2-443142193439"
prefix = "banana-ripeness-dataset/valid/overripe/"

# Create S3 client
s3 = boto3.client("s3")

# List objects in the folder
response = s3.list_objects_v2(
    Bucket=bucket,
    Prefix=prefix
)

# Print all image files
print("Images inside", prefix)
if "Contents" in response:
    for obj in response["Contents"]:
        print(obj["Key"])
else:
    print("No files found in this folder.")
```
